```python
# ============================================================
# Experiment 3: Reading Data from CSV, Excel, and Web
# ============================================================

import pandas as pd
import os

try:
    text_df = pd.read_csv("Google_data (2b.c1).csv")
    print("CSV File Loaded Successfully.\n")
except FileNotFoundError:
    print("CSV file not found!")
    text_df = pd.DataFrame()

try:
    excel_df = pd.read_excel("data (2c2).xlsx", sheet_name="Sheet1")
    print("Excel File Loaded Successfully.\n")
except FileNotFoundError:
    print("Excel file not found!")
    excel_df = pd.DataFrame()

web_df = pd.read_csv(
    "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
)

print("Web Data Loaded Successfully.\n")

if not text_df.empty:
    print("CSV Data")
    print(text_df.head())

if not excel_df.empty:
    print("\nExcel Data")
    print(excel_df.head())

print("\nWeb Data")
print(web_df.head())

if not text_df.empty:
    text_df.fillna(method="ffill", inplace=True)

if not excel_df.empty:
    excel_df.fillna(method="bfill", inplace=True)

web_df.dropna(inplace=True)

if not text_df.empty:
    text_df.to_csv("processed_text.csv", index=False)

if not excel_df.empty:
    excel_df.to_excel("processed_excel.xlsx", index=False)

print("\nProcessed files saved successfully!")
print("Experiment Completed Successfully.")
```
